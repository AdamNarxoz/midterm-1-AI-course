```python
import pandas as pd

ranks = pd.read_csv("my_data/retrieval_ranks.csv")

for config in ["A", "B"]:
    col = f"rank_{config}"
    print(f"\nConfig {config}")

    for k in [1, 3, 5]:
        hits = (ranks[col] <= k).sum()
        print(f"Recall@{k}: {hits}/60 = {hits/60:.4f}")

    mrr = (1 / ranks[col].clip(upper=10)).where(
        ranks[col] <= 10, 0
    ).mean()
    print(f"MRR@10: {mrr:.4f}")

print("\nPer-language recall")
for lang in ["EN", "RU", "KK"]:
    subset = ranks[ranks["lang"] == lang]
    for config in ["A", "B"]:
        col = f"rank_{config}"
        hits = (subset[col] <= 3).sum()
        print(lang, config, f"Recall@3 = {hits}/20 = {hits/20:.3f}")
```

```python
import math

def wilson(hits, n=60, z=1.96):
    p = hits / n
    d = 1 + z**2 / n
    centre = (p + z**2 / (2*n)) / d
    half = z * math.sqrt(
        p*(1-p)/n + z**2/(4*n**2)
    ) / d
    return centre-half, centre+half

for cfg in ["A", "B"]:
    for k in [3, 5]:
        hits = (ranks[f"rank_{cfg}"] <= k).sum()
        lo, hi = wilson(int(hits))
        print(
            f"{cfg}, k={k}: {hits}/60, "
            f"recall={hits/60:.4f}, "
            f"Wilson 95%=[{lo:.4f}, {hi:.4f}]"
        )

print("\nRecall by language for k=5")
for lang in ["EN", "RU", "KK"]:
    sub = ranks[ranks["lang"] == lang]
    hits = (sub["rank_B"] <= 5).sum()
    print(f"{lang}: {hits}/20 = {hits/20:.1%}")
```
